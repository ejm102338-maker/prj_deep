```python
import os
import pandas as pd
```

```python
# 1. 설정
folder_name = "종각젊음의거리"
source_dir = './data/datasets_local/' +  folder_name    # 원본 이미지 폴더
re_source_dir='./data/datasets_local/images'
output_csv = f'./data/datasets_local/image_list{folder_name}.csv'
prefix = 'img'               # 새로 붙일 이름의 접두사 (예: img_001)
valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp')

# 2. 파일 목록 가져오기
files = [f for f in os.listdir(source_dir) if f.lower().endswith(valid_extensions)]
files.sort()  # 이름순으로 정렬해서 순서대로 번호 부여

print(f"총 {len(files)}개의 파일을 재정의합니다...")
data_list = []
# 3. 루프 돌며 파일명 변경 및 데이터 수집
for i, old_name in enumerate(files, start=1):
    ext = os.path.splitext(old_name)[1]      # 확장자 추출 (.jpg 등)
    new_name = f"{folder_name}_{prefix}_{i:03d}{ext}"      # 새 이름 생성 (예: img_001.jpg)
    
    old_path = os.path.join(source_dir, old_name)
    new_path = os.path.join(re_source_dir, new_name)
    print(new_path)
    try:
        # 실제 파일 이름 변경 실행
        os.rename(old_path, new_path)
        
        # CSV에 들어갈 데이터 행 (딕셔너리 형태)
        data_list.append({
            "파일명": new_name,
            '문화유산':'0',
            '문화':'0',
            '자연':'0',
            '전시관':'0',
            '테마공원':'0',
            '시장':'0',
            '거리':'1',
            '공원':'0',
            '랜드마크' : folder_name
        })


    except Exception as e:
        print(f"오류 발생 ({old_name}): {e}")

# 4. CSV 저장
#df = pd.DataFrame(file_data)
df = pd.DataFrame(data_list)
# file_exists = os.path.isfile(output_csv)
# df.to_csv(output_csv, 
#             mode='a', 
#             index=False, 
#             header=not file_exists, 
#             encoding='utf-8-sig')

df.to_csv(output_csv, index=False, encoding='utf-8-sig')

print(f"✅ 작업 완료! '{output_csv}' 파일을 확인하세요.")
```
