```python
import os
import shutil

folder_path = r"C:\Users\user\Downloads\랜드마크 이미지\Training\서울특별시\[원천]서울특별시_009"

# place 리스트 파일
place_txt_path = r".\place.txt"



# place 리스트 읽기
with open(place_txt_path, "r", encoding="utf-8") as f:
    places = set(line.strip() for line in f if line.strip())



deleted = []
kept = []

for folder in os.listdir(folder_path):

    full_path = os.path.join(folder_path, folder)

    if not os.path.isdir(full_path):
        continue

    # base landmark 추출
    base_name = folder.split("_")[0]

    if base_name in places:

        kept.append(folder)

    else:

        print("삭제:", folder)

        shutil.rmtree(full_path)

        deleted.append(folder)



print("\n===== 결과 =====")
print("유지된 폴더:", len(kept))
print("삭제된 폴더:", len(deleted))



# 로그 저장
with open("kept_folders.txt","w",encoding="utf-8") as f:
    for k in kept:
        f.write(k+"\n")

with open("deleted_folders.txt","w",encoding="utf-8") as f:
    for d in deleted:
        f.write(d+"\n")
```
